## Random tensors

why random tensors?

Random tensors is important because many nn start from nueral networks and start struct something from these randoms numbers like diffusion models or gans 

we can do it my `torch.rand` : https://docs.pytorch.org/docs/stable/generated/torch.rand.html#torch.rand

```python
#create a random tensor of size (3,4)
random_tensor= torch.rand(1,3,4)
random_tensor
```

```python
random_tensor.ndim
```

```python
# create random tenosr with a shape  of (3,244,244) similar to an image
random_image_size_tensor=torch.rand(size =(3,244,244))
random_image_size_tensor.shape , random_image_size_tensor.ndim
```

## Tensor datatypes 
More about precision: https://en.wikipedia.org/wiki/Precision_(computer_science)

dtype : https://docs.pytorch.org/docs/stable/tensor_attributes.html#torch.dtype

**Note:** Tenosr datatypes is one of the 3 big erros you'll run into with PyTorch
1. Tenosrs not right datatype
2. Tenosrs not the right shape 
3. Tenosrs not on the right device   


```python
#float 32 tensor 
float_32_tensor= torch.tensor([.3,.4,.6] , dtype=None #data type for the tensor like float32 , float 16  do with precision and computing  
                              , device='cuda' ,  #where is the opperation will happned gpu or cpu 
                              requires_grad=False ) # track of the gradients with tensor opperations   
float_32_tensor
```

```python
float_32_tensor.dtype
```

```python
float_16_tenosr = float_32_tensor.type(torch.float16)
```

```python
float_16_tenosr
```

```python
float_16_tenosr * float_32_tensor #No error ? 
```

```python
int_32_tenosr = torch.tensor([3,4,5] , dtype=torch.int32 , device='cuda')
```

```python
int_32_tenosr * float_32_tensor # No error ? 
```
